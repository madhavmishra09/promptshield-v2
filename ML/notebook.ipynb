```python
from pathlib import Path
import pandas as pd

base_path = Path("dataset/Prompt-injection-dataset/core")

train_df = pd.read_parquet(base_path / "train-00000-of-00001.parquet")
val_df = pd.read_parquet(base_path / "validation-00000-of-00001.parquet")
test_df = pd.read_parquet(base_path / "test-00000-of-00001.parquet")

for name, data in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df),
]:
    print(f"\n{name}:")
    print("Shape:", data.shape)
    print("Labels:")
    print(data["label"].value_counts().sort_index())
    print("Unique groups:", data["group_id"].nunique())
    print("Duplicate texts:", data["text"].duplicated().sum())
```

```python
train_groups = set(train_df["group_id"])
val_groups = set(val_df["group_id"])
test_groups = set(test_df["group_id"])

print("Train ∩ Validation:", len(train_groups & val_groups))
print("Train ∩ Test:", len(train_groups & test_groups))
print("Validation ∩ Test:", len(val_groups & test_groups))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Build the baseline pipeline
model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            strip_accents="unicode",
            ngram_range=(1, 2),
            max_features=50000,
            sublinear_tf=True
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        )
    )
])

# Train using ONLY the training split
model.fit(
    train_df["text"],
    train_df["label"]
)

# Evaluate provisionally on validation data
val_predictions = model.predict(val_df["text"])
val_probabilities = model.predict_proba(val_df["text"])[:, 1]

print("VALIDATION RESULTS")
print("-" * 40)
print("Accuracy:", round(accuracy_score(
    val_df["label"], val_predictions
), 4))

print("Precision:", round(precision_score(
    val_df["label"], val_predictions
), 4))

print("Recall:", round(recall_score(
    val_df["label"], val_predictions
), 4))

print("F1:", round(f1_score(
    val_df["label"], val_predictions
), 4))

print("ROC-AUC:", round(roc_auc_score(
    val_df["label"], val_probabilities
), 4))

print("\nClassification Report:")
print(classification_report(
    val_df["label"],
    val_predictions,
    target_names=["Benign", "Injection"]
))

print("Confusion Matrix:")
print(confusion_matrix(
    val_df["label"],
    val_predictions
))
```

```python
import pandas as pd

error_df = val_df[["text", "label", "category", "source"]].copy()

error_df["predicted"] = val_predictions
error_df["probability"] = val_probabilities

false_positives = error_df[
    (error_df["label"] == 0) &
    (error_df["predicted"] == 1)
]

false_negatives = error_df[
    (error_df["label"] == 1) &
    (error_df["predicted"] == 0)
]

print("False positives:", len(false_positives))
display(
    false_positives[
        ["text", "category", "source", "probability"]
    ].head(10)
)

print("False negatives:", len(false_negatives))
display(
    false_negatives[
        ["text", "category", "source", "probability"]
    ].head(10)
)
```

```python
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support

category_results = val_df[
    ["text", "label", "category", "source"]
].copy()

category_results["predicted"] = val_predictions

rows = []

for category, group in category_results.groupby("category"):
    precision, recall, f1, _ = precision_recall_fscore_support(
        group["label"],
        group["predicted"],
        average="binary",
        zero_division=0
    )

    rows.append({
        "category": category,
        "samples": len(group),
        "missed_injections": int(
            ((group["label"] == 1) & (group["predicted"] == 0)).sum()
        ),
        "false_alarms": int(
            ((group["label"] == 0) & (group["predicted"] == 1)).sum()
        ),
        "precision": round(precision, 3),
        "recall": round(recall, 3),
        "f1": round(f1, 3)
    })

category_metrics = pd.DataFrame(rows)

display(
    category_metrics.sort_values(
        ["missed_injections", "false_alarms"],
        ascending=False
    )
)
```

```python
from sklearn.metrics import confusion_matrix
import pandas as pd

threshold_results = []

for threshold in [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70]:
    predictions = (val_probabilities >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        val_df["label"], predictions, labels=[0, 1]
    ).ravel()

    threshold_results.append({
        "threshold": threshold,
        "true_negatives": tn,
        "false_positives": fp,
        "false_negatives": fn,
        "true_positives": tp,
        "injection_recall": round(tp / (tp + fn), 4),
        "benign_specificity": round(tn / (tn + fp), 4),
        "accuracy": round((tn + tp) / (tn + fp + fn + tp), 4)
    })

display(pd.DataFrame(threshold_results))
```
